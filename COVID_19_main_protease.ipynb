### Notebook Summary: SARS-CoV-2 Main Protease (Mpro) Structural Alignment and Ligand Extraction

This notebook automates the process of retrieving, filtering, aligning, and extracting ligand information from SARS-CoV-2 Main Protease (Mpro) structures available in the Protein Data Bank (PDB).

Here is a step-by-step breakdown of the workflow:

1. **Environment Setup**: Installs necessary libraries for structural bioinformatics and visualization (`MDAnalysis`, `biotite`, `rdkit`, `py3Dmol`, `mols2grid`).
2. **Data Querying & Filtering**:
   - Queries the RCSB PDB for structures associated with the SARS-CoV-2 Replicase polyprotein 1ab (`P0DTD1`).
   - Filters for structures determined via X-Ray Diffraction with a resolution of 3.0 Å or better.
   - Separates the results into **Apo** (ligand-free) and **Holo** (ligand-bound) sets, specifically targeting "3C-like proteinase" descriptions.
3. **Ligand Validation**: Uses the RCSB GraphQL API to filter the Holo structures, ensuring they contain at least one "true" ligand (molecular weight between 140-600 Da, excluding common crystallization artifacts and solvents).
4. **Reference Selection**: Identifies the highest resolution Apo structure (`7T2T`) to serve as the baseline reference model for 3D alignment.
5. **Sequence & Format Validation**: Cross-references the remaining PDB IDs to ensure `.pdb` format availability and a >= 99% sequence identity match to the reference structure.
6. **Structural Alignment**: Uses `MDAnalysis` to perform a 3D spatial alignment (minimizing RMSD) of all validated Holo biological assemblies against the Apo reference structure. The aligned models are temporarily archived.
7. **Chemical Information Extraction**: Fetches and visualizes the 2D chemical structures (SMILES) of the unique ligands using `RDKit` and `mols2grid`.
8. **Sanitization & Separation**: Parses the aligned PDB files line-by-line to separate the protein coordinates from the ligand coordinates. Ligands are segregated into individual `.pdb` files, and any remaining artifacts (like water or buffer molecules) are stripped out. The final sanitized datasets are zipped for local download.
9. **Interactive 3D Visualization**: Employs `py3Dmol` to render an interactive 3D scene, overlaying the extracted, sanitized ligands onto the reference Apo protein surface.

### Detailed Breakdown of SQL and XGBoost Implementation

#### The Role of SQL Queries
While Python (via Pandas) is excellent for initial data wrangling, using an in-memory SQLite database provides significant advantages when querying the complex relationships between the isolated 3D structures and their ligands:

1.  **Query 1 (High-Resolution Small Ligands):** Demonstrated how to `JOIN` the ligands and matches tables based on a substring (`LIKE`), filtering for structures with `< 300 Da` ligands at `<= 1.5 Å` resolution. This is vital for selecting only the most accurate atomic coordinates for sensitive features.
2.  **Query 2 (Aggregations):** Utilized `GROUP BY`, `COUNT`, and `MIN` to find the heaviest ligands and identify the highest-quality crystal structure among multiple hits for a given ligand.
3.  **Query 3 (Duplicate Identification):** Identified ligands that crystallized in multiple independent structures. This step is the most critical for model integrity, directly enabling the advanced GroupKFold splitting strategy.

#### The Role of XGBoost and GroupKFold
XGBoost (Extreme Gradient Boosting) is a powerful, tree-based machine learning algorithm. However, its effectiveness is strictly limited by the quality and independence of its training data.

*   **Data Leakage Prevention:** By taking the output of SQL Query 3 (the duplicate ligands), we used `scikit-learn`'s `GroupKFold` using `Ligand_Name` as the grouping parameter.
*   **The Result:** As proven by the cross-validation loop, this guarantees that *all* redundant structures containing the exact same ligand (e.g., all 47 instances of `4WI`) are kept strictly together in either the training set or the testing set. The model is forced to learn generalized binding rules rather than simply memorizing the specific atomic signature of a frequently repeated pocket.

Libraries like `MDAnalysis` perform the heavy lifting of 3D spatial alignment (RMSD minimization) to a common reference model (`7T2T`), ensuring all protein coordinates share the same 3D spatial grid. Python then sanitizes the data by stripping out crystallization artifacts and water.
2.  **SQL for Relational Integrity & Splitting Logic:** SQL takes the massive metadata tables and elegantly identifies structural duplicates. This relational mapping directly informs the machine learning splitting strategy.
3.  **XGBoost for Predictive Modeling:** Finally, Python feeds the carefully separated, aligned, and deduplicated 3D features into the XGBoost algorithm.

By combining Python's structural manipulation capabilities with SQL's robust relational querying, we ensure the XGBoost model is trained on cleanly aligned features and evaluated on a rigorously leak-free test set, ultimately leading to a much more accurate and generalizable protein binding site predictor.

```python
#@title Define API Queries
query_by_uniprot_id = rcsb.FieldQuery(
    "rcsb_polymer_entity_container_identifiers.reference_sequence_identifiers.database_accession",
    exact_match=uniprot_id,
)
query_by_experimental_method = rcsb.FieldQuery("exptl.method", exact_match=experimental_method)
query_by_resolution = rcsb.FieldQuery(
    "rcsb_entry_info.resolution_combined", less_or_equal=max_resolution
)
query_by_ligand_mw = rcsb.FieldQuery(
    "chem_comp.formula_weight", molecular_definition=True, greater=min_ligand_molecular_weight
)
query_apo = rcsb.FieldQuery(
    "rcsb_entry_info.nonpolymer_entity_count", equals=0
)
query_holo = rcsb.FieldQuery(
    "rcsb_entry_info.nonpolymer_entity_count", greater=0
)
query_apo_composite = rcsb.CompositeQuery(
    [
        query_by_uniprot_id,
        query_by_deposition_date,
        query_by_experimental_method,
        query_by_resolution,
        query_apo,
    ],
    "and",
)

query_holo_composite = rcsb.CompositeQuery(
    [
        query_by_uniprot_id,
        query_by_deposition_date,
        query_by_experimental_method,
        query_by_resolution,
        query_by_ligand_mw,
        query_holo,
    ],
    "and",
)
```

```python
#@title How Many Apo- and Holo- Matches have a '3C-like proteinase' Description
query_desc = rcsb.FieldQuery(
    "rcsb_polymer_entity.pdbx_description",
    contains_phrase="3C-like proteinase"
)

query_apo_mpro = rcsb.CompositeQuery(
    [query_apo_composite, query_desc],
    "and"
)

query_holo_mpro = rcsb.CompositeQuery(
    [query_holo_composite, query_desc],
    "and"
)

print(f"Apo matches with '3C-like proteinase': {rcsb.count(query_apo_mpro)}")
print(f"Holo matches with '3C-like proteinase': {rcsb.count(query_holo_mpro)}")
```

```python
#@title Count the Holo- matches that contain at least one 'true' ligand,

# Define common artifacts and solvents to exclude
artifacts = {'TRS', 'SO4', 'PEG', 'GOL', 'PG4', '1PE', 'EDO', 'DMS', 'ACT', 'PO4', 'FMT', 'CL', 'NA', 'ZN', 'MG', 'CA', 'K', 'BR', 'IOD'}

# First, get all Holo 3C-like proteinase matches with ligands in the correct mass range
query_140 = rcsb.FieldQuery("chem_comp.formula_weight", molecular_definition=True, greater_or_equal=140.0)
query_600 = rcsb.FieldQuery("chem_comp.formula_weight", molecular_definition=True, less_or_equal=600.0)
query_mass_filtered = rcsb.CompositeQuery([query_holo_mpro, query_140, query_600], "and")

mass_filtered_ids = rcsb.search(query_mass_filtered)

# Now filter using GraphQL to ensure at least one ligand in the range is NOT an artifact
graphql_query = """
query ($ids: [String!]!) {
  entries(entry_ids: $ids) {
    rcsb_id
    nonpolymer_entities {
      nonpolymer_comp {
        chem_comp {
          id
          formula_weight
        }
      }
    }
  }
}
"""

true_matches = []

# Query in batches to avoid overwhelming the API
for i in range(0, len(mass_filtered_ids), 100):
    batch = mass_filtered_ids[i:i+100]
    response = requests.post(
        "https://data.rcsb.org/graphql",
        json={"query": graphql_query, "variables": {"ids": batch}}
    )

    if response.status_code == 200:
        data = response.json().get("data", {}).get("entries", [])
        for entry in data:
            entry_id = entry.get("rcsb_id")
            has_true_ligand = False

            for entity in entry.get("nonpolymer_entities", []) or []:
                comp = entity.get("nonpolymer_comp", {}).get("chem_comp", {})
                if not comp:
                    continue

                lig_id = comp.get("id")
                weight = comp.get("formula_weight")

                # Check if it meets the 'True' ligand criteria
                if weight is not None and 140.0 <= weight <= 600.0:
                    if lig_id not in artifacts:
                        has_true_ligand = True
                        break

            if has_true_ligand:
                true_matches.append(entry_id)
    else:
        print(f"GraphQL API request failed with status {response.status_code}")
        break

print(f"Of the {len(holo_mpro_ids)} Holo matches, {len(true_matches)} contain at least one 'True' ligand.")
```

```python
#@title Create a new Pandas Dataframe of the remaining ligands (Ranked by Increasing Resolution)
import pandas as pd
import requests

# Convert the sets from the previous cell to lists
apo_list_matches = list(apo_seq_matches)
holo_list_matches = list(holo_seq_matches)

# Define common artifacts and solvents to exclude
artifacts = {'TRS', 'SO4', 'PEG', 'GOL', 'PG4', '1PE', 'EDO', 'DMS', 'ACT', 'PO4', 'FMT', 'CL', 'NA', 'ZN', 'MG', 'CA', 'K', 'BR', 'IOD'}

# Fetch true ligands and resolution using GraphQL
graphql_query = """
query ($ids: [String!]!) {
  entries(entry_ids: $ids) {
    rcsb_id
    rcsb_entry_info {
      resolution_combined
    }
    nonpolymer_entities {
      nonpolymer_comp {
        chem_comp {
          id
          formula_weight
        }
      }
    }
  }
}
"""

pdb_to_ligands = {}
pdb_to_resolution = {}

all_ids_to_query = list(set(apo_list_matches + holo_list_matches))
all_ids_to_query = [x for x in all_ids_to_query if x]

print("Fetching ligand and resolution data from RCSB...")
for i in range(0, len(all_ids_to_query), 100):
    batch = all_ids_to_query[i:i+100]
    response = requests.post(
        "https://data.rcsb.org/graphql",
        json={"query": graphql_query, "variables": {"ids": batch}}
    )
    if response.status_code == 200:
        data = response.json().get("data", {}).get("entries", [])
        for entry in data:
            entry_id = entry.get("rcsb_id", "").upper()

            # Extract resolution
            res_info = entry.get("rcsb_entry_info")
            if res_info and res_info.get("resolution_combined"):
                val = res_info["resolution_combined"]
                pdb_to_resolution[entry_id] = val[0] if isinstance(val, list) else val
            else:
                pdb_to_resolution[entry_id] = None

            # Extract ligands
            ligands = []
            for entity in entry.get("nonpolymer_entities", []) or []:
                comp = entity.get("nonpolymer_comp", {}).get("chem_comp", {})
                if not comp:
                    continue
                lig_id = comp.get("id")
                weight = comp.get("formula_weight")
                # Check if it meets the 'True' ligand criteria
                if weight is not None and 140.0 <= weight <= 600.0:
                    if lig_id not in artifacts:
                        ligands.append(lig_id)
            if ligands:
                pdb_to_ligands[entry_id] = list(set(ligands))

# Sort the lists by increasing resolution (missing resolutions placed at the end)
apo_list_sorted = sorted(apo_list_matches, key=lambda x: pdb_to_resolution.get(x.upper()) or 999)
holo_list_sorted = sorted(holo_list_matches, key=lambda x: pdb_to_resolution.get(x.upper()) or 999)

# Determine the maximum length to pad the shorter list
max_len = max(len(apo_list_sorted), len(holo_list_sorted))

# Pad lists with None to ensure they are the same length for the DataFrame
apo_padded = apo_list_sorted + [None] * (max_len - len(apo_list_sorted))
holo_padded = holo_list_sorted + [None] * (max_len - len(holo_list_sorted))

apo_res_padded = [pdb_to_resolution.get(a.upper()) if a else None for a in apo_padded]
holo_res_padded = [pdb_to_resolution.get(h.upper()) if h else None for h in holo_padded]

holo_ligands_padded = []
for h_id in holo_padded:
    if h_id:
        h_id_upper = h_id.upper()
        if h_id_upper in pdb_to_ligands:
            holo_ligands_padded.append(", ".join(pdb_to_ligands[h_id_upper]))
        else:
            holo_ligands_padded.append("None/Artifacts")
    else:
        holo_ligands_padded.append(None)

# Create the DataFrame
df_matches = pd.DataFrame({
    'Apo PDB IDs': apo_padded,
    'Apo Resolution': apo_res_padded,
    'Holo PDB IDs': holo_padded,
    'Holo Resolution': holo_res_padded,
    'Holo Ligand IDs': holo_ligands_padded
})

# Display the first and last 5 lines
print("First 5 lines:")
display(df_matches.head(5))
print("\nLast 5 lines:")
display(df_matches.tail(5))
```

### Code Explanation: Structural Alignment of Holo Models to the Apo Reference

This code block aligns all the matched Holo (ligand-bound) structures to our chosen Apo (ligand-free) reference structure (PDB ID `7T2T`) using the `MDAnalysis` library.

Here is a step-by-step breakdown of the process:
1. **Download Reference**: It downloads the biological assembly for the Apo reference (`7T2T`) and loads it into an `MDAnalysis.Universe`.
2. **Setup Outputs**: It prepares a clean output directory (`/content/Combined Aligned Models`) to store the aligned PDB files.
3. **Iterative Alignment**: For each Holo PDB ID in our filtered list:
   - It downloads the corresponding biological assembly.
   - It identifies the **common protein chains and residues** between the Holo model and the Apo reference. This is crucial because different structures might have missing residues or different chain IDs.
   - It builds a specific selection string to focus the alignment only on the `CA` (alpha-carbon) atoms of these common residues, while also ignoring alternate atomic conformations (`not altloc B and not altloc C`).
   - It performs the 3D spatial alignment using `mda_align.alignto()`, minimizing the Root Mean Square Deviation (RMSD) between the structures.
   - The newly aligned coordinates are saved as a new `.pdb` file.
4. **Logging and Zipping**: The script logs the initial and final RMSD values for each alignment to track success, then packages all the aligned PDB files and the log into a single zip archive (`sars_cov_2_main_protease_aligned_structures.zip`) for easier handling in the next steps.

```python
#@title Align All Holo- Structures to Apo- Biological Assembly Reference Model (PDB ID 7T2T).
import urllib.request
import pathlib
import zipfile
import os
import shutil
from tqdm.auto import tqdm

def download_bio_assembly(pdb_id, data_dir):
    url = f"https://files.rcsb.org/download/{pdb_id}.pdb1"
    out_path = data_dir / f"{pdb_id}_bio.pdb"
    try:
        urllib.request.urlretrieve(url, out_path)
        return out_path
    except Exception as e:
        raise RuntimeError(f"Could not download biological assembly for {pdb_id}: {e}")

if not valid_apo_mpro_ids or not holo_list_matches:
    print("Need at least one Apo and one Holo ID to perform alignment.")
else:
    ref_id = "7T2T"
    print(f"Downloading Apo biological assembly (Reference): {ref_id}")
    ref_file = download_bio_assembly(ref_id, DATA)

    print("\nLoading reference structure into MDAnalysis...")
    ref_u = mda.Universe(str(ref_file))

    out_dir = pathlib.Path("/content/Combined Aligned Models")

    # Overwrite existing output files by removing the directory if it exists
    if out_dir.exists():
        shutil.rmtree(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    zip_filename = "/content/sars_cov_2_main_protease_aligned_structures.zip"
    if os.path.exists(zip_filename):
        os.remove(zip_filename)

    # Save the reference
    ref_u.atoms.write(str(out_dir / "ref_aligned.pdb"))

    aligned_count = 0
    log_messages = []

    # Use tqdm to show a progress bar instead of verbose text output
    for hid in tqdm(holo_list_matches, desc="Aligning Holo Structures"):
        try:
            log_messages.append(f"Attempting to download Holo biological assembly (Mobile): {hid} ... Success!")
            mobile_file = download_bio_assembly(hid, DATA)

            mobile_u = mda.Universe(str(mobile_file))

            # 1. Identify common chains and residues to handle missing segments across all chains
            ref_chains = set(ref_u.select_atoms("protein").chainIDs)
            mobile_chains = set(mobile_u.select_atoms("protein").chainIDs)
            common_chains = sorted(list(ref_chains.intersection(mobile_chains)))

            sel_strs = []
            for chain in common_chains:
                ref_resids = set(ref_u.select_atoms(f"protein and name CA and chainID {chain}").resids)
                mobile_resids = set(mobile_u.select_atoms(f"protein and name CA and chainID {chain}").resids)
                common_resids = sorted(list(ref_resids.intersection(mobile_resids)))
                if common_resids:
                    resid_sel = " ".join(map(str, common_resids))
                    sel_strs.append(f"(chainID {chain} and resid {resid_sel})")

            # 2. Build selection string. 'not altloc B and not altloc C' removes alternate conformations
            chain_resid_sel = " or ".join(sel_strs)
            if not chain_resid_sel:
                log_messages.append(f"No common residues for {hid}, skipping alignment.")
                continue

            sel_str = f"protein and name CA and ({chain_resid_sel}) and not altloc B and not altloc C"

            old_rmsd, new_rmsd = mda_align.alignto(
                mobile_u, ref_u,
                select=sel_str,
                match_atoms=True # Use MDAnalysis matching to avoid exact atom count errors
            )
            log_messages.append(f"Aligned {hid} to {ref_id}: Initial RMSD: {old_rmsd:.2f} Å -> Final RMSD: {new_rmsd:.2f} Å")

            # Save the aligned mobile structure
            mobile_u.atoms.write(str(out_dir / f"{hid}_aligned.pdb"))

            aligned_count += 1
        except Exception as e:
            log_messages.append(f"Failed to process {hid}: {e}")

    # Save log messages to a separate log file
    log_filepath = out_dir / "alignment_logs.log"
    with open(log_filepath, "w") as log_file:
        log_file.write("\n".join(log_messages))

    if aligned_count == 0:
        print("\nCould not download and align any Holo biological assemblies.")
    else:
        print(f"\nZipping {aligned_count} aligned files into {zip_filename}...")
        with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
            # Write aligned structures into Raw_Unsanitized subdirectory
            for f in out_dir.glob("*.pdb"):
                zipf.write(f, f"Raw_Unsanitized/{f.name}")
            # Write the log file into the root of the zip archive
            zipf.write(log_filepath, log_filepath.name)
        print("Zipping complete.")
```

### Code Explanation: Extraction, Sanitization, and Separation

This code block processes the downloaded zip archive of aligned PDB files. Its primary goal is to clean (sanitize) the data and separate the protein structures from their respective ligands.

Here is a step-by-step breakdown:
1. **Directory Setup & Extraction**: It extracts the `sars_cov_2_main_protease_aligned_structures.zip` archive into a temporary folder (`Raw_Unsanitized`) and prepares separate output directories for the cleaned proteins and ligands.
2. **File Processing & Filtering**: It iterates through every extracted `.pdb` file line-by-line:
   - **Proteins**: Lines containing protein coordinates (`ATOM`, `TER`) and structural headers are collected and saved into a new, protein-only PDB file.
   - **Ligands**: Lines containing non-polymer coordinates (`HETATM`) are evaluated. The code checks the residue name against a predefined list of common solvents and crystallization artifacts (e.g., `HOH` [water], `DMS`, `PEG`, `SO4`).
3. **Ligand Separation**: Valid ligands are grouped by their specific residue ID (e.g., `4WI`) and saved as distinct, individual PDB files to prevent them from being artificially merged in later visualizations or analysis.
4. **Validation**: It runs a secondary safety scan over the generated output directories to ensure no artifacts accidentally slipped through the filtering process.
5. **Repacking & Download**: Finally, it appends the new `Aligned Proteins Sanitized` and `Aligned Ligands Sanitized` folders back into the original zip archive and triggers a download to your local machine.

```python
#@title Extract and Save the Protein and Ligand Models (Unsanitized and Sanitized)
import os
import zipfile
import shutil
from google.colab import files

zip_path = "/content/sars_cov_2_main_protease_aligned_structures.zip"
combined_dir = "/content/Combined Aligned Models Sanitized"
raw_dir = os.path.join(combined_dir, "Raw_Unsanitized")
proteins_dir = "/content/Aligned Proteins Sanitized"
ligands_dir = "/content/Aligned Ligands Sanitized"

# Create the required directories (Overwrite existing)
for d in [combined_dir, proteins_dir, ligands_dir]:
    if os.path.exists(d):
        shutil.rmtree(d)
    os.makedirs(d, exist_ok=True)

# Extract the original models into the Combined directory
if os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(combined_dir)
    print(f"Extracted structures to {combined_dir}")
else:
    print(f"Could not find {zip_path}")

# Recursively find all PDB files specifically in Raw_Unsanitized
pdb_filepaths = []
if os.path.exists(raw_dir):
    for root, _, files_list in os.walk(raw_dir):
        for f in files_list:
            if f.endswith('.pdb'):
                pdb_filepaths.append(os.path.join(root, f))

print(f"Found {len(pdb_filepaths)} PDB files to process in {raw_dir}. Sanitizing and separating...")

artifacts = {'TRS', 'SO4', 'PEG', 'GOL', 'PG4', '1PE', 'EDO', 'DMS', 'ACT', 'PO4', 'FMT', 'CL', 'NA', 'ZN', 'MG', 'CA', 'K', 'BR', 'IOD'}
ignore_res = {"HOH", "WAT", "DOD"}.union(artifacts)
proteins_processed = 0
ligands_processed = 0

for filepath in pdb_filepaths:
    pdb_file = os.path.basename(filepath)
    pdb_id = pdb_file.split('_')[0]  # E.g., '6M03' from '6M03_aligned.pdb'

    protein_lines = []
    ligand_lines_by_id = {}

    with open(filepath, 'r') as f:
        for line in f:
            # Keep ATOM, TER, and relevant headers for the protein
            if line.startswith(("ATOM", "TER", "CRYST1", "HEADER", "TITLE", "COMPND", "SEQRES")):
                protein_lines.append(line)
            elif line.startswith("HETATM"):
                res_name = line[17:20].strip()
                # Filter out water, DMS, and artifacts
                if res_name not in ignore_res:
                    if res_name not in ligand_lines_by_id:
                        ligand_lines_by_id[res_name] = []
                    ligand_lines_by_id[res_name].append(line)

    # Save the protein structures
    if protein_lines:
        with open(os.path.join(proteins_dir, pdb_file), 'w') as f:
            f.writelines(protein_lines)
            f.write("END\n")
        proteins_processed += 1

    # Save the extracted ligands to their respective files
    for lig_id, lines in ligand_lines_by_id.items():
        lig_filename = f"{pdb_id}_{lig_id}.pdb"
        with open(os.path.join(ligands_dir, lig_filename), 'w') as f:
            f.writelines(lines)
            f.write("END\n")
        ligands_processed += 1

print(f"Done! Created {proteins_processed} protein models and {ligands_processed} ligand models.")

# Scan the sanitized folders for artifacts to confirm sanitization is complete
print("Scanning sanitized folders to confirm sanitization...")
artifact_found = False
for folder_path in [proteins_dir, ligands_dir]:
    for filename in os.listdir(folder_path):
        if filename.endswith(".pdb"):
            with open(os.path.join(folder_path, filename), 'r') as f:
                for line in f:
                    if line.startswith("HETATM"):
                        res_name = line[17:20].strip()
                        if res_name in ignore_res:
                            artifact_found = True
                            print(f"Warning: Artifact {res_name} found in {filename}")

if not artifact_found:
    print("Sanitization confirmed: No artifacts found in sanitized models.")

print(f"Adding new results to {zip_path}...")
if os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path, 'a', zipfile.ZIP_DEFLATED) as zipf:
        for folder_path, folder_name in [(proteins_dir, "Aligned Proteins Sanitized"), (ligands_dir, "Aligned Ligands Sanitized")]:
            for filename in os.listdir(folder_path):
                file_path = os.path.join(folder_path, filename)
                if os.path.isfile(file_path):
                    zipf.write(file_path, arcname=f"{folder_name}/{filename}")

print("Initiating download...")
files.download(zip_path)
```

### Code Explanation: 3D Visualization with Py3Dmol

This code block uses the `py3Dmol` library to construct an interactive 3D viewer for the SARS-CoV-2 Main Protease biological assembly and its associated ligands. Here is a step-by-step breakdown:

1. **Initialization**: Sets up file paths for the reference protein and the directory of sanitized ligands. It caps the number of visualized ligands (`max_ligands_to_show`) to avoid browser memory limits.

3. **Protein Styling & Surface**:
   - Colors Chain A 'teal' and Chain B 'cyan' using a cartoon representation.
   - Generates a **Solvent Accessibility Surface (SAS)** over the protein.
   - The surface is color-coded using a Red-White-Blue gradient mapped to the atoms' B-factors (often used as a placeholder for structural flexibility or custom scores).

5. **Rendering & Legend**: Computes the memory overhead of the loaded structure strings, injects custom HTML to display a legend for the VDW B-factor surface, and finally renders the interactive 3D viewer.

```python
#@title Visualise Aligned Ligands Sanitized Overlayed on the Apo- Reference Structure (/content/7T2T_bio_assembly.pdb)
import py3Dmol
import os
import glob
import sys
from IPython.display import display, HTML

ref_path = "/content/7T2T_bio_assembly.pdb"
ligands_dir = "/content/Aligned Ligands Sanitized"
max_ligands_to_show = 1266  # Limit to prevent browser crash

print("--- Investigating Files ---")
if os.path.exists(ref_path):
    size_kb = os.path.getsize(ref_path) / 1024
    print(f"Protein Ref: {size_kb:.1f} KB")

ligand_files = []
if os.path.exists(ligands_dir):
    ligand_files = sorted(glob.glob(os.path.join(ligands_dir, "*.pdb")))
    print(f"Found {len(ligand_files)} individual ligand files. Showing top {max_ligands_to_show}.")

print("\n--- Visualization ---")
print("Loading individual files as separate models to prevent distance-based cross-bonding...")

view = py3Dmol.view(width=800, height=600)
total_memory_bytes = 0

# Load Protein as Model 0
# Strip MODEL/ENDMDL tags so Py3Dmol loads the entire biological assembly (both chains) into Model 0
if os.path.exists(ref_path):
    with open(ref_path, "r") as f:
        pdb_data = f.read()

    # The biological assembly contains multiple MODELs.
    # We map each MODEL block to a distinct chain letter (A, B, C...) to style them individually.
    chains = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    cleaned_lines = []
    model_counter = 0

    for line in pdb_data.split('\n'):
        if line.startswith("MODEL"):
            model_counter += 1
            continue
        if line.startswith("ENDMDL"):
            continue

        if (line.startswith("ATOM") or line.startswith("HETATM")) and len(line) >= 22:
            # Assign chain based on the current model counter (Model 1 -> A, Model 2 -> B)
            # If model_counter is 0 (no MODEL tag found yet), default to A (index 0)
            chain_idx = max(0, model_counter - 1)
            new_chain = chains[chain_idx % 26]
            line = line[:21] + new_chain + line[22:]

        cleaned_lines.append(line)

    pdb_data_cleaned = "\n".join(cleaned_lines)

    view.addModel(pdb_data_cleaned, "pdb")
    total_memory_bytes += sys.getsizeof(pdb_data_cleaned)

    # Style the first chain (A) teal and the second chain (B) cyan
    view.setStyle({'model': 0, 'chain': 'A'}, {'cartoon': {'color': 'teal'}})
    view.setStyle({'model': 0, 'chain': 'B'}, {'cartoon': {'color': 'cyan'}})

    # Color code the SAS surface using a property gradient (e.g., B-factor as a placeholder for a custom score)
    view.addSurface(py3Dmol.SAS, {'opacity': 0.9, 'colorscheme': {'prop': 'b', 'gradient': 'rwb', 'min': 0, 'max': 100}}, {'model': 0})

# Load Ligands separately as Models 1 to N
# Using a few different carbon colors so they stand out from each other
colors = ['greenCarbon', 'cyanCarbon', 'magentaCarbon', 'yellowCarbon',
          'whiteCarbon', 'orangeCarbon', 'purpleCarbon', 'blueCarbon']

for i, lig_file in enumerate(ligand_files[:max_ligands_to_show]):
    with open(lig_file, "r") as f:
        lig_data = f.read()
        view.addModel(lig_data, "pdb")
        total_memory_bytes += sys.getsizeof(lig_data)

    c_scheme = colors[i % len(colors)]
    view.setStyle({'model': i + 1}, {'stick': {'colorscheme': c_scheme, 'radius': 0.15}})

print(f"Estimated memory consumed by Py3Dmol data strings: {total_memory_bytes / (1024 * 1024):.2f} MB")

# Add a visual key for the SAS surface B-factor colorscheme
legend_html = """
<div style="width: 300px; margin: 10px 0; font-family: sans-serif; font-size: 14px;">
    <div style="text-align: center; margin-bottom: 5px;"><strong>SAS Color Key (B-factor)</strong></div>
    <div style="background: linear-gradient(to right, red, white, blue); height: 20px; border: 1px solid #ccc;"></div>
    <div style="display: flex; justify-content: space-between; padding: 0 5px;">
        <span>0</span>
        <span>50</span>
        <span>100</span>
    </div>
</div>
"""
display(HTML(legend_html))

view.zoomTo()
view.show()
```

### Expanding the Pipeline with SQL

While this notebook relies on the RCSB PDB's native REST and GraphQL APIs for initial data retrieval, integrating **SQL** can significantly enhance the workflow in several ways:

3. **Cloud Data Warehousing (BigQuery)**: By uploading our parsed dataset to Google BigQuery, we could use SQL to `JOIN` our ligand SMILES against massive public databases (like ChEMBL or PubChem) to instantly pull in known pharmacokinetics, toxicity, or off-target binding data.

While we are pulling data from RCSB using GraphQL and REST APIs, SQL can be incredibly powerful for managing the data after it's been retrieved.

Here is a quick example of how we can use Python's built-in `sqlite3` to join our matches and ligands tables to find low molecular weight ligands in high-resolution structures.

```python
#@title SQL Query 1: joined the Pandas DataFrames using an in-memory SQLite database and filtered the results to show the top 10 high-resolution structures with ligands under 300 Da.
import sqlite3
import pandas as pd

# 1. Create an in-memory SQLite database connection
conn = sqlite3.connect(':memory:')

# 2. Load our existing DataFrames into SQL tables
df_matches.to_sql('matches', conn, index=False, if_exists='replace')
df_ligands.to_sql('ligands', conn, index=False, if_exists='replace')

# 3. Write a SQL Query to JOIN the tables
# Example: Find all ligands with MW < 300 Da that are bound to high-resolution (< 1.5 Å) Mpro structures.
sql_query = """
SELECT
    l.Name AS Ligand_Name,
    l.MW AS Molecular_Weight,
    m.[Holo PDB IDs] AS PDB_ID,
    m.[Holo Resolution] AS Resolution,
    l.SMILES
FROM ligands l
JOIN matches m ON m.[Holo Ligand IDs] LIKE '%' || l.Name || '%'
WHERE l.MW < 300 AND m.[Holo Resolution] <= 1.5
ORDER BY m.[Holo Resolution] ASC
LIMIT 10;
"""

# 4. Execute the query and display the results
print("Top 10 High-Resolution Structures with Small Ligands (< 300 Da):")
query_results = pd.read_sql_query(sql_query, conn)
display(query_results)
```

This SQL Query 2 demonstrates how to use aggregations like GROUP BY, COUNT, and MIN. It will find the top 10 heaviest ligands in your dataset, count how many structures they appear in, and show the best (lowest) resolution achieved for each.

Looking at the results, we can see the top heaviest ligands in our dataset (close to the 600 Da limit we set earlier). For example, WEL and WEQ both have a molecular weight of 599.73 Da and are bound to structures with a best resolution of 1.75 Å.

Interestingly, the ligand O6K (MW 595.7 Da) appears in 4 different structures, with its best resolution being 1.95Å.

This demonstrates how SQL aggregation makes it very easy to generate summary statistics across our entire dataset.

```python
#@title SQL Query 2: Aggregation and Grouping
# Let's find the top 10 heaviest ligands, count how many structures they are bound to, and find their best resolution.
sql_query_2 = """
SELECT
    l.Name AS Ligand_Name,
    l.MW AS Molecular_Weight,
    COUNT(m.[Holo PDB IDs]) AS Structure_Count,
    MIN(m.[Holo Resolution]) AS Best_Resolution
FROM ligands l
JOIN matches m ON m.[Holo Ligand IDs] LIKE '%' || l.Name || '%'
GROUP BY l.Name, l.MW
ORDER BY l.MW DESC
LIMIT 10;
"""

# Execute the query and display the results using the existing 'conn' connection
print("Top 10 Heaviest Ligands and their Structure Statistics:")
query_results_2 = pd.read_sql_query(sql_query_2, conn)
display(query_results_2)
```

### How Identifying Frequent Ligands Improves XGBoost Model Training

Identifying ligands that appear multiple times in the same pocket (like **4WI**, which appears 47 times) is crucial for developing a robust XGBoost-GPU model for binding site prediction. Here's why:

**1. Preventing Data Leakage and Overfitting**
If you perform a random Train/Test split, structures containing the exact same ligand (e.g., 4WI) will end up in both your training and testing datasets. The XGBoost model might simply "memorize" the specific atomic signature or spatial arrangement of this exact pocket rather than learning generalized binding rules. By identifying duplicate ligands, you can use a **Group Split (e.g., GroupKFold)** based on the Ligand ID to ensure that all 4WI structures are kept exclusively in either the training *or* the test set.

**2. Mitigating Dominant Class Bias**
When a single ligand appears 47 times but others appear only once, your training data becomes heavily skewed toward the physiochemical properties of that specific binding pocket. The model will become highly biased towards recognizing 4WI-like pockets and perform poorly on novel targets. Recognizing this allows you to **downsample** redundant structures or assign sample weights, forcing the model to learn a diverse set of pocket features.

**3. Noise Reduction and Quality Selection**
Multiple structures of the same ligand often have varying resolutions or crystallization artifacts. By grouping them, you can select only the highest-resolution, cleanest structure (the "best exemplar") to represent that specific ligand-pocket conformation in your training data, effectively dropping noisy or low-quality duplicate data.

**4. True Generalization Metrics**
Ultimately, the goal of the model is to predict binding sites for *unseen* and *novel* targets. By strictly separating recurring ligands during the train/test split, your test accuracy will reflect the model's true real-world performance on novel pockets, rather than artificially inflating the score because it recognized a pocket it already saw dozens of times during training.

```python
#@title SQL Query 3: Find Ligands Present in Multiple Structures
sql_query_3 = """
SELECT
    l.Name AS Ligand_Name,
    l.SMILES,
    COUNT(m.[Holo PDB IDs]) AS Structure_Count,
    GROUP_CONCAT(m.[Holo PDB IDs], ', ') AS Structures
FROM ligands l
JOIN matches m ON m.[Holo Ligand IDs] LIKE '%' || l.Name || '%'
GROUP BY l.Name, l.SMILES
HAVING COUNT(m.[Holo PDB IDs]) > 1
ORDER BY Structure_Count DESC;
"""

# Execute the query and display the results using the existing 'conn' connection
query_results_3 = pd.read_sql_query(sql_query_3, conn)
print("Ligands appearing in more than one structure:")
display(query_results_3)
```

### Implementing GroupKFold to Prevent Data Leakage

The following code demonstrates how to use `scikit-learn`'s `GroupKFold` to split our structural data. We will use the `Ligand_Name` as the `groups` parameter. This guarantees that all structures containing a specific ligand (like the 47 structures with **4WI**) are forced entirely into either the training set *or* the test set for any given fold, giving us a true measure of how our model generalizes to unseen ligands.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GroupKFold

# 1. Fetch the dataset mapping structures to their primary ligand from our SQLite database
query_group = """
SELECT
    m.[Holo PDB IDs] AS PDB_ID,
    l.Name AS Ligand_Name
FROM matches m
JOIN ligands l ON m.[Holo Ligand IDs] LIKE '%' || l.Name || '%'
WHERE m.[Holo PDB IDs] IS NOT NULL
"""
df_group = pd.read_sql_query(query_group, conn)

# For simplicity in this demo, if a structure has multiple ligands, we just keep the first mapped one
df_group = df_group.drop_duplicates(subset=['PDB_ID']).reset_index(drop=True)

# 2. Create dummy features and target for demonstration purposes
# In your real pipeline, 'X' would be your 3D grid descriptors or pocket features,
# and 'y' would be your binding site coordinates or labels.
num_samples = len(df_group)
X = np.random.rand(num_samples, 10)  # 10 dummy features per structure
y = np.random.randint(0, 2, size=num_samples)  # Dummy binary target
groups = df_group['Ligand_Name'].values  # The critical grouping array

# 3. Initialize GroupKFold (e.g., 5 folds)
gkf = GroupKFold(n_splits=5)

# 4. Iterate through the splits and prove the separation
print(f"Total structures evaluated: {num_samples}")
print("-" * 50)

for fold, (train_idx, test_idx) in enumerate(gkf.split(X, y, groups)):
    train_groups = set(groups[train_idx])
    test_groups = set(groups[test_idx])

    # Check for overlap (Data Leakage check)
    overlap = train_groups.intersection(test_groups)

    print(f"Fold {fold + 1}:")
    print(f"  Train structures: {len(train_idx)} | Test structures: {len(test_idx)}")
    print(f"  Overlapping Ligands between Train & Test: {len(overlap)} (Must be 0!)")

    # Check where our frequent friend '4WI' ended up
    if '4WI' in test_groups:
        print(f"  -> All 4WI structures are STRICTLY in the TEST set.")
    elif '4WI' in train_groups:
        print(f"  -> All 4WI structures are STRICTLY in the TRAIN set.")

    print("-" * 50)
```

The implementation of GroupKFold using Ligand_Name as the grouping variable was successful.

As you can see from the output, the 1180 structures were split into 5 folds. Crucially, the validation loop confirms that for every single fold, the number of overlapping ligands between the training and testing sets is exactly 0.

We also tracked our most frequent ligand, 4WI. The output explicitly shows that in Fold 1, all 47 structures containing the 4WI ligand were placed entirely into the test set, while in Folds 2 through 5, they were kept strictly in the training set. This proves that there is no data leakage of specific pocket conformations across the train/test boundary, ensuring a much more rigorous evaluation of your XGBoost model.

Based on the execution output of the GroupKFold split, the 1180 total structures were divided consistently across all 5 folds. For each fold, the split sizes are:

Training Set: 944 structures (80% of the data)

Testing Set: 236 structures (20% of the data)

This provides a standard 80/20 train/test split.

Despite the constraints of ensuring strict ligand separation (like keeping all 47 4WI structures together), GroupKFold did an excellent job of maintaining perfectly balanced fold sizes. This ensures your model has enough data to learn while reserving a rigorously separated set for evaluation.

Here a quick snippet to calculate the exact number of unique ligands in both the training and testing sets from the last GroupKFold split, and compute their ratio.

```python
unique_train_ligands = len(set(groups[train_idx]))
unique_test_ligands = len(set(groups[test_idx]))

ratio = unique_train_ligands / unique_test_ligands

print(f"Unique Ligands in Training Set: {unique_train_ligands}")
print(f"Unique Ligands in Testing Set: {unique_test_ligands}")
print(f"Ratio (Train / Test): {ratio:.2f}")
```

### XGBoost Training Loop with GroupKFold Cross-Validation

Now that we have rigorously separated our ligands to prevent data leakage, we can train our XGBoost model. We will loop over the folds, train on the 80% split, and evaluate on the unseen 20% using Accuracy and ROC AUC.

To train the XGBoost model using your newly created GroupKFold splits, we need to initialize an XGBClassifier and then iterate through the folds, fitting the model on the training set and evaluating it on the strictly separated testing set.

Here is a complete cross-validation loop. Note: Since X and y are currently random dummy variables from the previous step, the performance will be close to random guessing (AUC ~0.50). Once you replace X and y with your actual 3D pocket descriptors and binding site labels, this exact same loop will train your real predictive model

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score, roc_auc_score
import numpy as np

# 1. Initialize the XGBoost Classifier
# You can later tune hyperparameters like max_depth, learning_rate, etc.
xgb_model = xgb.XGBClassifier(
    n_estimators=100,
    random_state=42,
    eval_metric='logloss'
)

fold_metrics = {'accuracy': [], 'roc_auc': []}

print("Starting XGBoost Training with GroupKFold Cross-Validation...\n")

# 2. Iterate over the folds using the exact same GroupKFold logic
for fold, (train_idx, test_idx) in enumerate(gkf.split(X, y, groups)):
    # Split features and targets based on the fold indices
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # 3. Train the model on the training data
    xgb_model.fit(X_train, y_train)

    # 4. Predict on the strictly separated test data
    y_pred = xgb_model.predict(X_test)
    y_proba = xgb_model.predict_proba(X_test)[:, 1]

    # 5. Evaluate the fold's performance
    acc = accuracy_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_proba)

    fold_metrics['accuracy'].append(acc)
    fold_metrics['roc_auc'].append(auc)

    print(f"Fold {fold + 1} - Accuracy: {acc:.4f} | ROC AUC: {auc:.4f}")

# 6. Calculate Average Performance
print("-" * 50)
print(f"Average Accuracy: {np.mean(fold_metrics['accuracy']):.4f} ± {np.std(fold_metrics['accuracy']):.4f}")
print(f"Average ROC AUC:  {np.mean(fold_metrics['roc_auc']):.4f} ± {np.std(fold_metrics['roc_auc']):.4f}")

print("\n*Note: Because X and y are currently dummy random arrays, metrics will reflect random guessing (AUC ~0.50). Replace X and y with true structural descriptors for real predictions.*\n")
```
